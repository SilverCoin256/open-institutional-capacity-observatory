# Research Showcase

This notebook summarizes the v1 observatory surfaces for a research audience.

```python
from statistics import mean
from oico.datasets import build_all
from oico.io import ROOT, read_csv

assert build_all()['status'] == 'pass'
queue = [r for r in read_csv(ROOT / 'datasets' / 'processed' / 'queue_observations.csv') if r['qai']]
asi = read_csv(ROOT / 'datasets' / 'processed' / 'asi_scores.csv')
print('Mean observed QAI:', round(mean(float(r['qai']) for r in queue), 3))
print('Mean ASI score:', round(mean(int(r['total_asi_score']) for r in asi), 3))
print('Use these as descriptive summaries, not causal claims.')
```
