# Reproduce And Inspect

This notebook rebuilds the OICO release candidate and inspects core table counts.

```python
from oico.datasets import build_all
from oico.io import ROOT, read_csv

report = build_all()
assert report['status'] == 'pass'
summary = {}
for name in ['institutions', 'queue_observations', 'institutional_indicators', 'asi_scores']:
    summary[name] = len(read_csv(ROOT / 'datasets' / 'processed' / f'{name}.csv'))
print(summary)
```
